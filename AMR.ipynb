# Part B

```python
cols_B = [
    'phenotype-species',
    'phenotype-antibiotic_name',
    'phenotype-resistance_phenotype',
    'phenotype-collection_year',
    'phenotype-country',
    'phenotype-geographical_region',
    'phenotype-BioSample_ID']

df_B = df[cols_B].copy()
df_B = df_B.dropna()
display(df_B)
drug = ["tetracycline","streptomycin", "gentamicin",'chloramphenicol',' erythromycin','azithromycin','methicillin',' trimethoprim' ,'amikacin' ,'kanamycin']
antibio_pheno_country = df_B.groupby(["phenotype-antibiotic_name","phenotype-resistance_phenotype","phenotype-geographical_region"]).agg(Count=("phenotype-antibiotic_name","count"))
b = pd.DataFrame(antibio_pheno_country.reset_index())
b = b[b['phenotype-antibiotic_name'].isin(drug)]
display(b)
```

```python
b["Total"] = b.groupby([
    "phenotype-antibiotic_name",
    "phenotype-geographical_region"
])["Count"].transform("sum")

# Percentage of each resistance phenotype
b["Percentage"] = b["Count"] / b["Total"] * 100
ax = plt.figure(figsize=(14,5))

g = sns.catplot(
    data=b,
    x="phenotype-geographical_region",
    y="Percentage",
    hue="phenotype-resistance_phenotype",
    col="phenotype-antibiotic_name",
    col_wrap=4,
    kind="bar",
    palette="rocket",
    height=3.5,
    sharey=True
)

g.set_axis_labels("Region", "Percentage (%)")
g.set_titles("{col_name}")

for ax in g.axes.flat:
    ax.tick_params(axis="x", rotation=45)

plt.show()
```

```python
cols_C = [
    'phenotype-BioSample_ID',
    'phenotype-isolate',
    'phenotype-species',
    'phenotype-antibiotic_name',
    'phenotype-resistance_phenotype',
    'phenotype-collection_year',
    'phenotype-country',
    'phenotype-isolation_source_category',
    'phenotype-host_age',
    'phenotype-host_sex'
]

df_C = df[cols_C].copy()
# display(df_C)
df_C["phenotype-collection_year"]= df_C["phenotype-collection_year"].dropna().astype(int)
df_C["phenotype-collection_year"] = df_C["phenotype-collection_year"].sort_values()
df_C_R = df_C[df_C["phenotype-resistance_phenotype"]=="resistant"]
# x = df_C_R["phenotype-species"].value_counts()
top_sp = ['Mycobacterium tuberculosis', 'Klebsiella pneumoniae',
'Escherichia coli',
'Salmonella enterica',
'Acinetobacter baumannii',
'Streptococcus pneumoniae',
'Staphylococcus aureus',
'Neisseria gonorrhoeae',
'Shigella sonnei',
'Campylobacter jejuni']
# display(x)
df_C_R_S = df_C_R[df_C_R["phenotype-species"].isin(top_sp)]
year = df_C_R_S.groupby(["phenotype-species","phenotype-collection_year"]).agg(counts=("phenotype-collection_year","count"))  
year = year.reset_index()
display(year)
sns.lineplot(data= year, x="phenotype-collection_year", y = "counts", hue = "phenotype-species",palette="rocket")
plt.xlim(1980,2020)
plt.legend(prop={
    'style': 'italic',      # 'normal', 'italic', or 'oblique'
    'weight': 'light',       # 'normal', 'bold', 'heavy', 'light'
    'size': 10              # numeric size or 'small', 'medium', 'large'
})
plt.show()
```

```python
amr = pd.read_csv("/Users/pp/Downloads/amr_records-2.csv")
amr_r = amr[amr["pheno_geno_merged-resistance_phenotype"]=="resistant"]
amr_r
```

```python
d = amr_r[["pheno_geno_merged-species",'pheno_geno_merged-isolation_source_category','pheno_geno_merged-geographical_subregion']]
d.loc[d["pheno_geno_merged-geographical_subregion"]=="Latin America and the Caribbean"] = "Latin America"
display(d)

# Top 10 bacterial species
top_species = d["pheno_geno_merged-species"].value_counts().head(7).index
top_species
d = d[d["pheno_geno_merged-species"].isin(top_species)]

# Count unique sample-species observations by region
distribution = pd.crosstab(
    d["pheno_geno_merged-species"],
    d["pheno_geno_merged-geographical_subregion"]
)
distribution

# Percentage within each bacterial species
distribution_pct = distribution.div(
    distribution.sum(axis=0), axis=1
) * 100

plt.figure(figsize=(11, 7))

sns.heatmap(
    distribution_pct,
    annot=True,
    fmt=".1f",
    cmap="rocket_r",
    cbar_kws={"label": "Share of recorded isolates (%)"}
)

plt.title("Geographical Distribution of Bacterial Isolates\n", fontsize = 20)
plt.xlabel("Geographical Region")
plt.ylabel("Bacterial Species")
plt.yticks(fontstyle="italic")
plt.tight_layout()
plt.show()
```

```python
plot_data = distribution_pct.T
display(plot_data)

# Plot
fig, ax = plt.subplots(figsize=(12, 8))

colors = sns.color_palette("rocket", n_colors=len(plot_data.columns))

plot_data.plot(
    kind="barh",
    stacked=True,
    color=colors,
    ax=ax,
    alpha = 0.9,
    width=0.8
)

ax.set_xlabel("Share of recorded bacterial isolates (%)")
ax.set_ylabel("Geographical Region")
ax.set_xlim(0, 100)

ax.legend(
    title="Bacterial Species",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

ax.invert_yaxis()
sns.despine()
ax.tick_params(axis="x", labelsize=14)
ax.tick_params(axis="y", labelsize=14)
ax.legend(
    title="Bacterial Species",
    prop={"style": "italic", "size": 14},
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)
ax.set_xlabel("Percentage (%)", fontsize=16)
ax.set_ylabel("Bacterial Species", fontsize=16)

plt.show()
```
